# Adding the Risk Levels and storing them

```python
Customers = ["Ahmed", "Sara", "Omar", "Mona","Ali", "Nour", "Youssef", "Hana","Khaled", "Mariam", "Hassan", "Laila"]

Scores = [91, 45, 78, 66, 88, 30, 72, 95, 51, 60, 84, 39]

Risk_Levels = []

High = 0
Medium = 0
Low = 0
Total = 0
for customer, score in zip(Customers,Scores):
    if score >= 85:
        risk = "High"
        High += 1
    elif 60 <= score < 85:
        risk = "Medium"
        Medium += 1
    else:
        risk = "Low"
        Low += 1
    
    Risk_Levels.append(risk)
    Total += score

Average = Total / 12


print(f"{High} , {Low} , {Medium} , {Total} , {Average:.2f}, {Risk_Levels}")        
```
