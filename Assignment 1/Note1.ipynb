# Task 1

## Maps

Input:	    User's current location, destination, traffic data, road conditions, and travel history.

AI Task:	Predict traffic conditions and estimate the fastest route to a destination.

Output:	    Suggested routes, estimated arrival time, and traffic information.

Risk 1:	    Incorrect traffic predictions may lead users to take longer routes.

Risk 2:	    Location tracking may create privacy and data security concerns.

## Shopping

Input:	    Customer searches, browsing history, previous purchases, ratings, and product information.

AI Task:	Analyze customer preferences and recommend relevant products and best recommendations.

Output:	    Personalized product recommendations and suggestions.

Risk 1:	    Recommendations may be biased and limit product variety.

Risk 2:	    Collecting customer data may create privacy concerns.

## Banking

Input:	    Account information, transaction amount, recipient details, and transaction history.

AI Task:	Analyse data, check user's transaction history and dedicect if there is any fraud actions.

Output:	    accepted transaction or refused transaction.

Risk 1:	    Incorrect fraud detection may flag legitimate transactions.

Risk 2:	    Financial data leaks could expose users to fraud and identity theft.
