# The Four Rules Of Bulding An Ai System:

## 1.Data Analyst

Excels in Excel, SQL, Power Bl, Python EDA. Builds executive dashboards and extracts business insights. Baisiclly descibes how the model works and what should it do on papers.

## 2.Data Scientist

Focuses on statistics, hypothesis testing, A/B experiments, predictive ML models, and business storytelling. Baisiclly creates the prototype modle and runs it until it's ready to be added to the system.

## 3.ML Engineer

Bridges models to production: training pipelines, REST APIs, model latency, Docker, and drift monitoring. Baisiclly puts the the model created by the Data scientis and puts it int the system.

## 4.Al Engineer

Architects GenAl systems: prompt engineering, RAG over knowledge bases, tool-calling agents, and evaluation. it takes the whole system and prepares with all the requierments for the useage of the customer.
